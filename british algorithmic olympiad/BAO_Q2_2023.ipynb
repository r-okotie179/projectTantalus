```python
def PET(time, position):
    '''
    finding angle
    '''
    # input time is in nanoseconds
    time = (time)
    ind = time.index(min(time))
    closest_pos = position[ind]
    dictionary = {"A":0.25, "B":0.5, "C":0.75} # dictionaries are a more efficient way than a bunch of if/else statements
    conv_pos = (closest_pos[0] - 1) + dictionary.get(closest_pos[1])
    if conv_pos < 0:
        conv_pos = dictionary.get(closest_pos[1])
    angle = (conv_pos / 32) * 360
    angle = format(angle, ",.2f")

    '''
    finding distance
    '''
    min_time = min(time) * (10**-9)
    c = 3*(10**8)
    l = c * min_time
    d = (0.45 - l) * 100 # centimetres
    d = format(d, ",.2f")
    print(angle, d)

time = [1.000, 2.000]
position = [[20, "B"], [4, "B"]]

PET(time, position)
```
